```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

embeddings
```

```python
from sklearn.metrics.pairwise import cosine_similarity
text="Hello, Bhargav love Sahithya"
text2="Hello, Sahithya love Bhargav"

emb=embeddings.embed_query(text)
emb2 = embeddings.embed_query(text2)
print(f"Text: {text}")
print(f"Embedding length:{len(emb)}")
print(emb)
print(cosine_similarity([emb],[emb2]))

dot_product = np.dot(emb,emb2)
norm_a=np.linalg.norm(emb)
norm_b=np.linalg.norm(emb2)
print(dot_product/(norm_a*norm_b))
```

```python
res=[]
def semantic_search(query,docs,embder):
    query_emb=embder.embed_query(query)
    doc_emb=embder.embed_documents(docs)
    for i,doc in enumerate(doc_emb):
        res.append((cosine_similarity([query_emb],[doc]),docs[i]))
    res.sort(reverse=True)
    return res
docu=[
    "sahithya loves bhargav",
    "bhargav love sahithya",
    "ajay dislikes sahithya",
]

que="who is sahithya"
print(semantic_search(que,docu,embeddings))
```
